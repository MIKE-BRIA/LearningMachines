## Python logging


#### Logging is a crucial aspect of any application, providing a way to track events, errors and operational information. Python inbuilt logging module offers a flexible framework for emitting log messages from python programs

```python
import logging

##Configure basic logging settings
logging.basicConfig(level=logging.DEBUG)

#log messages
logging.debug('This is a debug message')
logging.info('This is an info message')
logging.warning('This is a warning message')
logging.error('This is a error message')
logging.critical('This is a critical message')
```

```python
import logging

### Configuring the logging

logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

logging.debug('This is a debug message')
logging.info('This is an info message')
logging.warning('This is a warning message')
logging.error('This is a error message')
logging.critical('This is a critical message')
```
