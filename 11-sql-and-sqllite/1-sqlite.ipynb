```python
## Sql and Sqlite

import sqlite3
```

```python
### Connect to a sqlite database

connection=sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## creating a table

cursor.execute('''
         Create Table If Not Exists employees(
             id Integer Primary Key,
             name Text Not null,
             age integer,
             department text
         )
''')

connection.commit()
```

```python
cursor.execute('''select * from employees''')
```

```python
### Inserting data in sqlite table

cursor.execute('''
               Insert into employees(name,age,department)
               values('Brian',23,'Machine Learning')
''')


cursor.execute('''
               Insert into employees(name,age,department)
               values('Luckas',23,'Robotics')
''')


cursor.execute('''
               Insert into employees(name,age,department)
               values('Michael',23,'Finance')
''')


cursor.execute('''
               Insert into employees(name,age,department)
               values('Owen',23,'Aerospace')
''')


connection.commit()
```

```python
### Query data from sqlite table

cursor.execute('select * from employees')

rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute('''
               update employees
               set age = 34
               where name = 'Owen'
''')


connection.commit()
```

```python
cursor.execute('select * from employees')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
### Deleting data from the table

cursor.execute('''
               delete from employees
               where name = 'Owen'
               ''')

connection.commit()
```

```python
cursor.execute('''
               select *
               from employees
               ''')


rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
##Working with sales Data

connection = sqlite3.connect('sale_data.db')
cursor = connection.cursor()


cursor.execute('''
               create table if not exists sales(
                   id integer primary key,
                   date text not null,
                   product text not null,
                   sales integer,
                   region text
               )
               ''')


sales_data=[
    ('2023-01-01', 'Product1', 100,'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product3', 150,'West'),
    ('2023-01-04', 'Product4', 205, 'South'),
    ('2023-01-05', 'Product5', 700,'North'),
    ('2023-01-06', 'Product6', 280, 'East')

]


cursor.executemany('''
                   insert into sales(date,product,sales,region)
                   values(?,?,?,?)
                   ''',sales_data)

connection.commit()
```

```python
cursor.execute('''
               select *
               from sales
               ''')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
### Closing the connection

connection.close
```
