## File operation - Read and write files

```python
### Read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## Read a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) # .strip() removes the newline character at the end of each line
```

```python
## Write to a file - overwrites the existing content of the file with the new content

with open('example.txt', 'w') as file:
    file.write('A new world order is in the making. \n')
    file.write('And we are here to dominate it. \n')
```

```python
## Write a file without overwriting the existing content of the file

with open('example.txt', 'a') as file:
    file.write('We are appending a new line to the file. \n')
```

```python
### Writing a list of lines to a file

lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```
