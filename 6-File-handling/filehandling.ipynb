## File operation - Read and write files

```python
### Read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## Read a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) # .strip() removes the newline character at the end of each line
```

```python
## Write to a file - overwrites the existing content of the file with the new content

with open('example.txt', 'w') as file:
    file.write('A new world order is in the making. \n')
    file.write('And we are here to dominate it. \n')
```

```python
## Write a file without overwriting the existing content of the file

with open('example.txt', 'a') as file:
    file.write('We are appending a new line to the file. \n')
```

```python
### Writing a list of lines to a file

lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']

with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
## Binary files - bin files are files that contain data in a format that is not human-readable. They are often used for images, audio, and other types of media.


## Writing to a binary file

data = b'\x00\x01\x02\x03\x04\x05'

with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
## Reading from a binary file

with open('example.bin', 'rb') as file:
    binary_data = file.read()
    print(binary_data)
```

```python
## Read the content from a source text file and write to a destination text file

with open('example.txt', 'r') as source_file:
    content = source_file.read()
    print('source content: \n', content)


with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)

with open('destination.txt', 'r') as file:
    print('final destination file content: \n',file.read())
```

```python
##  Read a text file and count the number of lines, words, and characters .
# Counting lines,words and characters in a file


def count_file_content(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)

        print(f'Lines: {line_count}')
        print(f'Words: {word_count}')
        print(f'Characters: {char_count}')

count_file_content('example.txt')
```

```python
## writing and then reading a file

with open('example.txt', 'w+') as file:
    file.write('This is a test file.\nIt contains multiple lines of text.\nThis is the third line.')
    file.seek(0) # Move the cursor back to the beginning of the file to be able to read the content of the file after writing to it.
    content = file.read()
    print(content)
```
