```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt


#loading data
dataset= load_dataset('lukebarousse/data_jobs')
df=dataset['train'].to_pandas()


#Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_usa = df[df['job_country'] == 'United States'].copy()
df_usa.sample(8)
```

```python
df_usa['job_posted_month']=df_usa['job_posted_date'].dt.strftime('%B')
df_usa['job_posted_month'].dropna()
```

```python



df_usa_pivot
# Convert index to Series, then use .dt accessor
df_usa_pivot['month_no'] = pd.to_datetime(df_usa_pivot.index.to_series(), format='%B').dt.month
df_usa_pivot = df_usa_pivot.sort_values('month_no')
df_usa_pivot
```
