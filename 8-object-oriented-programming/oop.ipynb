```python
### Modeling a bank account

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print(f'{amount} deposited. New balance is {self.balance}')

    def withdraw(self,amount):
        if amount>self.balance:
            print('Insufficient funds')
        else:
            self.balance-=amount
            print(f'{amount} withdrawn, New account balance is {self.balance}')

    def get_balance(self):
        return self.balance



## Creating an object of the class BankAccount
account1=BankAccount('Alice',10000)
print(account1.balance)
```

```python
account1.deposit(5000)
```

```python
account1.withdraw(2000)
```

```python
account1.get_balance()
```

## Inheritence

#### Inheritence is a fundamental concept in oop that allows a class to inherit attributes and methods from another class

```python
## Inheritence

## Parent class

class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f'The person will drive the {self.enginetype} car')

```

```python
car1=Car(2,4,'petrol')
car1.drive()
```

```python
## child class

class ElectricCar(Car):
    def __init__(self,windows,doors,enginetype,battery_capacity,is_self_driving):
        super().__init__(windows,doors,enginetype)
        self.battery_capacity=battery_capacity
        self.is_self_driving=is_self_driving

    def battery_status(self):
        print(f'the battery capacity of the car is {self.battery_capacity} kWh and it is {'self driving' if self.is_self_driving else 'not self driving'}')

    def selfdriving(self):
        print(f'the car supports self driving: {self.is_self_driving}')

    def self_driving_status(self):
        if self.is_self_driving:
            print('the car is self driving')
        else:
            print('The car is not self driving')
```

```python
Tesla=ElectricCar(4,4,'electric',98,True)
Tesla.self_driving_status()
Tesla.battery_status()
Tesla.selfdriving()
Tesla.drive()
```

```python
### Multiple inheritence

## when a class inherits from more than one parent class,

class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print('Subclasses should implement this method')

## Base class 2
class Pet:
    def __init__(self,owner):
        self.owner = owner

## Derived class inheriting from both Animal and Pet
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)

    def speak(self):
        return f'{self.name} says woof!'



dog=Dog('Bosko','Alice')
dog.speak()
print(f'{dog.name} is owned by {dog.owner}')
print(dog.speak())
```

### Polymorphism

#### Allows objects of different classes to be treated as objects of a common superclass 

```python
## Method overriding
## allows a child class to provide a specific implementation of a method that is already defined in its parent class.



##Base class

class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        return 'Sounds are made by animals'

##Derived class
class Dog(Animal):
    def speak(self):
        return f'{self.name} says woof!'


class Cat(Animal):
    def speak(self):
        return f'{self.name} says meow!'


dog=Dog('Buddy')
print(dog.speak())
cat=Cat('Whiskers')
print(cat.speak())
```

```python
### Polymorphism with functions and methods

##Base class
class Shape:
    def area(self):
        return 'The area of the shape is calculated using the formula specific to the shape'


## Derived class 1
class Rectangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height


    def area(self):
        return self.width*self.height

## Derived class 2
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius**2


## Function that demonstrates polymorphism

def print_area(shape):
    print(f'The area of the shape is: {shape.area()}')


rect=Rectangle(6,19)
circle=Circle(7)

print_area(rect)
print_area(circle)
```

```python
### Polymorphism with abstract base classes - are used to define a common methods for a group of related objects, they can enforce that derived classes implement particular methods, promoting consistency across different implementations.
```

```python
from abc import ABC,abstractmethod

## Abstract base class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

## Derived class 1
class Car(Vehicle):
    def start_engine(self):
        return 'Car engine is starting with a key'

## Derived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return 'Motorcycle engine is starting with a button'

##Function that demonstrates polymorphism

def start_vehicle_engine(Vehicle):
    print(Vehicle.start_engine())


##Creating objects of car and motorcycle
car=Car()
bike=Motorcycle()

start_vehicle_engine(car)
start_vehicle_engine(bike)
```

 ### Encapsulation and Abstraction

 #### Encapsulation and Abstraction help is designing robust,maintainable and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features

```python
### Encapsulation - is the concept of wrapping data(variables) and methods(functions) together as a single unit. it restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of Data.
```

```python
##Encapsulation with getter and setter methods
##Public,protected and private variables

class Person:
    def __init__(self,name,age):
        self.name=name ## Public variable, can be accessed directly
        self.age=age ##Public variable, can be accessed directly


def get_name(person):
    return person.name

person=Person('Allitta',45)
get_name(person)
```

```python
##Private Variables examples
##Private variables are only accessible within the class they are defined in and are not accessible from outside the class.

class Person:
    def __init__(self,name,age,weight,gender):
        self.name=name
        self.__age=age ##Private variable, cannot be accessed directly outside the class
        self.__gender=gender ##Private variable, cannot be accessed directly outside the class
        self.weight=weight


## this is for public variable,which can be accessed directly outside the class
def get_name(person):
    return person.name

person=Person('Michael',30,70,'Male')
print(get_name(person))


##Private variables cannot

def get_age(person):
    return person.__age

print(get_age(person))
```

```python
class Person:
    def __init__(self,name,age,weight,gender,location,school):
        self.name=name
        self.__age=age ##Private variable, cannot be accessed directly outside the class
        self.__gender=gender ##Private variable, cannot be accessed directly outside the class
        self.weight=weight
        self._location=location ##Protected variable, can be accessed within class and derived classes but not outside the class
        self._school=school ##Protected variable, can be accessed within class and derived classes but not outside the class


class Employee(Person):
    def __init__(self,name,age,weight,gender,location,school,employee_id,department):
        super().__init__(name,age,weight,gender,location,school)
        self.employee_id=employee_id
        self.department=department



employee=Employee('John',28,80,'Male','New York','NYU',12345,'IT')

print(employee._location)
```

```python
##Encapsulation with getter and setter methods


class Person:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age


    ##getter method for name
    def get_name(self):
        return self.__name

    ##setter method for name
    def set_name(self,name):
        self.__name=name

    ##getter method for age
    def get_age(self):
        return self.__age

    ##setter method for age
    def set_age(self,age):
        if age <0:
            print('Age cannot be a negative number')
        else:
            self.__age=age



person=Person('Alice',30)

##Accessing and modifying private variables using getter and setter methods

print(person.get_name())
print(person.get_age())


person.set_name('Bobclark')
print(person.get_name())

person.set_age(35)
print(person.get_age())

person.set_age(-5)
```

```python
##Abstraction - is the concept of hiding the complex implementation details and showing only the necessary features of an object. it helps to reduce complexity and allows the user to focus on interactions at a higher level without needing to understand the underlying details.
```

```python
##Abstract classes and methods
##Abstract classes are classes that cannot be instantiated and are meant to be subclassed. They can contain abstract methods, which are methods that are declared but not implemented in the abstract class. Subclasses of the abstract class must provide an implementation for the abstract methods.


from abc import ABC,abstractmethod

class Vehicle(ABC):
    def drive(self):
        print('The person will drive the vehicle')

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print('Car engine is starting with a key ignition')



def operate_vehicle(machine):
    machine.start_engine()

car=Car()

operate_vehicle(car)
```

### Magic Methods


##### Magic methods are special methods in Python that have double underscores on both sides (__like_this__) and are automatically called by Python behind the scenes.

```python
## Basic Methods

class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

person1=Person('Alice',30)
print(person1)
```

```python
##overriding the __str__ method to provide a more informative string representation of the object

class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):
        return f'{self.name} is {self.age} years old.'


person =Person('Alice',30)
print(person)
```
