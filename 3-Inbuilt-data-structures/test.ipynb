## SETS

```python
## sets are unordered collection of unique items. they are defined using curly braces {} or the set() constructor. sets do not allow duplicate elements and do not maintain any specific order.


## creating a set

my_set = {1, 2, 3, 4, 5}
print(my_set)
```

```python
set_from_tuples=set((tuples))
```

```python
print(set_from_tuples)
```

```python
## Set operations

## adding and removing elements from a set

my_set={1, 2, 3, 4, 5}
my_set.add(6)
print(my_set)
```

```python
my_set.remove(3)
```

```python
print(my_set)
```

```python
## Pop method removes and returns an arbitrary element from the set. if the set is empty, it raises a KeyError.
popped_element = my_set.pop()
print(f"Popped element: {popped_element}")
print(f"Remaining set: {my_set}")
```

```python
## clear all the elements from the set

my_set.clear()
print(my_set)
```

```python
## set memebership test

my_set = {1, 2, 3, 4, 5}
print(3 in my_set)
print(6 in my_set)
print(type(my_set))
```

```python
## mathematical set operations

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Union

union_set = set1.union(set2)

print(union_set)

## Intersection - Gives us back results that are common between the two variables

intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6,7}
set2 = {4,5,6,7,8,9}

## Difference - gives us back unique data that are not in both the two variales
print(set1.difference(set2))

set1.difference_update(set2)
print(set1)
```

```python
## Symmetric difference - gives us both unique elements from both the variables
set1 = {1,2,3,4,5,6,7}
set2 = {4,5,6,7,8,9}
set1.symmetric_difference(set2)
```

```python
## Set Method

set1 = {1,2,3,4}
set2 = {3,4,5,6}

## is subset -- checks if the values in one variable are present in another variable

print(set1.issubset(set2))

## is superset
set1.issuperset(set2)
```

```python
## counting unique words in text

text = 'I love what am doing right now an i also love myself'

words = text.split()
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```

## Dictionaries

```python
## dictionaries are unordered collections of items. they store data in key-value pairs. Keys must be unique and immutable(e.g, Strings, numbers or tuples) while values can be for any type

## Creating dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
student={'name':"Brian Michael",'Age':23,"Grade":30}
print(student)
```

```python
type(student)
```

```python
##accessing dictionary elements

print(student)
print(student['Grade'])
student['Age']
```

```python
## Accessing using the get method

print(student.get('name'))
```

```python
## Modifying dictionaries elements

print(student)
student['Age']=32
student['Occupation']='Full Time Trader and Founder'
print(student)
```

```python
## delete key

del student['Grade']
print(student)
```

```python
## Dictionary Methods

keys = student.keys() ## for getting all the keys
print(keys)
values = student.values() ##get all values
print(values)

items = student.items() ##get all key value pairs
print(items)
```

```python
## Shallow copy
student_copy = student
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## enables you to make a copy of the current dictionary and you can make changes to it without affecting the original dictionary
print(student_copy1)
print(student)
student['name']='Shaniz wendy'
print(student_copy1)
print(student)
```

```python
## Iterating over dictionaries
## Using loops

#iterating over keys
for keys in student.keys():
    print('keys: ',keys)

#iterate over values
for i in student.values():
    print('values: ',i)



#iterate over key value pairs

for i,y in student.items():
    print(f"{i}:{y}")
```

```python
## Nested Dictionaries

students={
    'student1':{'name':'Owino','age':23,'major':'Medical lab sciences','year_of_study':4},
    'student2':{'name':'Rhodgers','age':25,'major':'Acturial Science','year_of_study':2}
}

print(students)
```

```python
# Access nested dic elements

print(students['student2']['name'])
```

```python
for student_id,student_info in students.items():
    print(f'{student_id}:{student_info}')
    for key,value in student_info.items():
        print(f'{key}:{value}')
```

```python
## Dictionary comprehension

squares={x:x**2 for x in range(15)}
print(squares)
```

```python
## conditional dic comprehension

even = {x:x**4 for x in range(7) if x%2==0}
print(even)
```

```python
## merging two dic into one

dict1 = {'a':1,'b':2}
dict2 = {'b':4,'c':9}
merged_dic = {**dict1,**dict2}
print(merged_dic)
```
