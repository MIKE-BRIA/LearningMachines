## Functions

#### function is a block of code that performs a specific task, Functions help in organizing code, reusing code and improving readability

```python
## Syntax
def function_name(parameters):
    """Docstring"""
    #function body
    return expression
```

```python
## Function to check if a number is even or odd

def even_or_odd(num):
    """This function finds even or odd numbers"""

    if num %2 ==0:
        print('the number is even')
    else:
        print('the number is odd')
```

```python
## calling the function

even_or_odd(46)
```

```python
## function with multiple parameters

def add(a,b):
    """this function add two values together"""
    c = a+b
    return c

results = add(100,3)
print(results)
```

```python
## Default parameters

def greet(name='Guest'):
    """Greets People when they come in the company"""
    print(f'Hello {name} welcome to Gimanx group of companies')


greet('Michael Brian')
```

```python
## Variable length arguments
## positional and keywords arguments

##when you have many parameters --- postitional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,'Brian')
```

```python
## Keywords arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_details(name='Brian Michael',age='23',country='Kenya')
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f'Positional arguments: {val}')

    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_details(1,2,3,'Michael',country='America')
```
