## Functions

#### function is a block of code that performs a specific task, Functions help in organizing code, reusing code and improving readability

```python
## Syntax
def function_name(parameters):
    """Docstring"""
    #function body
    return expression
```

```python
## Function to check if a number is even or odd

def even_or_odd(num):
    """This function finds even or odd numbers"""

    if num %2 ==0:
        print('the number is even')
    else:
        print('the number is odd')
```

```python
## calling the function

even_or_odd(46)
```

```python
## function with multiple parameters

def add(a,b):
    """this function add two values together"""
    c = a+b
    return c

results = add(100,3)
print(results)
```

```python
## Default parameters

def greet(name='Guest'):
    """Greets People when they come in the company"""
    print(f'Hello {name} welcome to Gimanx group of companies')


greet('Michael Brian')
```

```python
## Variable length arguments
## positional and keywords arguments

##when you have many parameters --- postitional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,'Brian')
```

```python
## Keywords arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_details(name='Brian Michael',age='23',country='Kenya')
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f'Positional arguments: {val}')

    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_details(1,2,3,'Michael',country='America')
```

```python
## Return statement

def multiply(a,b):
    return a*b,a


multiply(2,4)
```

### Function usecase examples

```python
## Temperature conversion

def convert_temp(temp,unit):
    """This function converts temperature between celsius and fehrenheit"""

    if unit=='C':
        return temp * 9/5 + 32 ##converting celsius to fehr
    elif unit == 'F':
        return (temp - 32)*5/9
    else:
        return None


print(convert_temp(34,'F'))
print(convert_temp(65,'C'))
```

```python
## Password strength checker

def is_strong_password(password):
    """checks if the password is strong"""

    if len(password)< 8:
        return False

    if not any(char.isdigit() for char in password):
        return False

    if not any(char.islower() for char in password):
        return False

    if not any(char.isupper() for char in password):
        return False

    if not any(char in '!@#$%&+' for char in password):
        return False

    return True



print(is_strong_password('MMgweno@23'))
print(is_strong_password('owneisagosdk'))
print(is_strong_password('dhdj23762932'))
```

```python
## Calculate the total cost of items in a shopping cart

def calculate_total_cost(cart):
    """This function calculate the cost of all items in the cart"""

    total_cost =0
    for item in cart:
        total_cost+=item['price']*item['quantity']

    return total_cost


## Example of cart data

cart=[
    {'name':'Apple','price':2.2,'quantity':4},
    {'name':'Mango','price':1.2,'quantity':6},
    {'name':'laptop','price':982.2,'quantity':1},
    {'name':'bag','price':32.2,'quantity':2},
    {'name':'monitor','price':1000.2,'quantity':2},
    {'name':'shirt','price':5,'quantity':4}
]


calculate_total_cost(cart)
```

### Lambda function

###### lambda functions are small anonymous functions defined using lambda keyword. They can have any number of arguments but only one expression. they are commonly used for short operations or as arguments to higher-order functions

```python
## syntax
lambda arguments:expression
```

```python
## normal function
def addition(a,b):
    return a+b

addition(7,8)
```

```python
##lambda function

addition = lambda a,b:a+b

print(addition(8,9))
```

```python
def even(num):
    if num%2==0:
        return True

even(64)
```

```python
even1 = lambda num:num%2==0

even1(64)
```

```python
## map functions --- map() = applies a function to all items in a list

numbers=[4,6,8,1,2,3,6,9]

def square(number):
    """calculates the square of a number"""

    return number**2

square(6)
```

```python
list(map(lambda X:X**2, numbers))
```

### Map() function in python

##### the map() function applies a given function to all items in an input list(or any other iterable)and returns a map object(on iterator). this is particularly useful for transforming data in a list comprehensively

```python
## calculating the square of numbers in a list using map and lambda function

numbers=[9,6,7,8,1,2,3,12,45]

list(map(lambda x:x*x, numbers))
```

```python
## using map with normal functions

def square(x):
    return x*x


list(map(square,numbers))
```

```python
### Map multiple iterables

numbers1=[1,2,3,4]
numbers2=[5,6,7,8]

addition=list(map(lambda x,y:x+y, numbers1,numbers2))
print(addition)
```

```python
## use map to convert a list of string to integers

str_numbers = ['1','2','3','4','5','6']

int_numbers=list(map(int,str_numbers))

print(int_numbers)
```

```python

words=['mango','banana','michael','founder','robotics','smartkid']

upper_words=list(map(str.upper, words))
print(upper_words)
```

```python
## Using map with dictionaries

def get_name(person):
    return person['name']

people=[
    {'name':'Brian','age':23},
    {'name':'Michael','age':15}
]

list(map(get_name,people))
```

```python
list(map(lambda x:x['name'], people))
```

## filter () function

#### the Filter() function constructs an iterator from elements of an iterable for which a function returns true. it is used to filter out items from a list(or any other iterable) based on a condition

```python
def even(num):
    if num%2==0:
        return True

lst=[1,3,2,4,6,8,5,90,33,12,41,56,86,41]
list(filter(even,lst))
```

```python
list(filter(lambda num:num>7,lst))
```

```python
list(filter(lambda num:num%2==1 ,lst))
```

```python
## filter with lambda function and multiple conditions

even_and_greater_than_five = list(filter(lambda x:x>5 and x%2==0,lst))
print(even_and_greater_than_five)
```

```python
## filter with dictionaries
## check if age is greater than 25

people=[
    {'name':'Brian','age':23},
    {'name':'Michael','age':15},
    {'name':'Owen','age':23},
    {'name':'Mitchelle','age':35},
    {'name':'Bridgit','age':26},
    {'name':'Mourice','age':30},
]

def age_past25(person):
    return person['age']>25


list(filter(age_past25,people))
```

```python
list(filter(lambda person:person['age']>21, people))
```
