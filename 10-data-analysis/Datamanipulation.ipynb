## Data Manipulation and Analysis with Pandas

#### Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform and extract insights from data. 

```python
import pandas as pd
```

```python
df=pd.read_csv('sales_data.csv')

df.head(5)
```

```python
df.tail(5)
```

```python
df.describe()
```

```python
df.dtypes
```

```python
## Handling missing values

df.isnull().any()
```

```python
df.isnull().sum()
```

```python
df_filled=df.fillna(0)
```

```python
#### filling missing values with the mean of the column

df['Profit_FillNA']=df['profit'].fillna(df['profit'].mean())
df
```

```python
df.dtypes
```

```python
### Renaming columns

df=df.rename(columns={'Profit_FillNA':'Profit_filled_data'})
df.head(10)
```

```python
### Applying a function to one of the columns

df['New Value']=df['profit'].apply(lambda x:x*1.67)
df.head(10)
```

```python
## Data aggregating and grouping

df.head(6)
```

```python
grouped_mean = df.groupby('country')['profit'].mean()
```

```python
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['state','country'])['profit'].sum()
print(grouped_sum)
```

```python
## Aggregate multiple functions

grouped_agg=df.groupby(['state','country'])['profit'].agg(['mean','sum','count'])
grouped_agg
```
