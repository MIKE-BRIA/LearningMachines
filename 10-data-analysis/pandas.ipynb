### Pandas-DataFrame and Series

#### Pandas is a powerful data manipulation library in Python , widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a dataframe is a two-dimensional, size-mutable, and potentially hererogeneous tabular data structure with labeled axes (rows and columns)

```python
import pandas as pd
```

```python
##Series

data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
```

```python
## Creating a series from a dictionary

data={'a':1,'b':2,'c':3}
series=pd.Series(data)
print(series)
```

```python
data=[10,20,30,40,50]

index=['a','b','c','d','e']
pd.Series(data,index=index)
```

```python
##DataFrame
## Creating a DataFrame from a dictionary

data={
    'Name':['Alice','Bob','Charlie'],
    'Age':[25,30,35],
    'City':['New York','Los Angeles','Chicago']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a dataframe from a list of dictionaries

data=[
    {'Name':'Alice','Age':25,'City':'New York'},
    {'Name':'Bob','Age':30,'City':'Los Angeles'},
    {'Name':'Charlie','Age':35,'City':'Chicago'}
]

df=pd.DataFrame(data)
print(df)
```
