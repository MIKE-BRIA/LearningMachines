### Data visualization with matplotlib


#### Matplotlib enables creation of static,animated and interactive visualizations. it is widely used for data visualization in data science and analytics

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]


##Create a line plot
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic line plot')
plt.show()
```

```python

from matplotlib.lines import lineStyles


x=[1,2,3,4,5]
y=[1,4,9,16,25]


## Create a customized line plot

plt.plot(x,y, color='red',linestyle='-.',marker='*',linewidth=1,markersize=6)
plt.grid(True)
```

```python
# Multiple Plots

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]


plt.figure(figsize=(9,5))

plt.subplot(1,2,1)
plt.plot(x,y1,color='green')
plt.title('Plot One')



plt.subplot(1,2,2)
plt.plot(x,y2,color='green')
plt.title('Plot Two')

```

```python
# Multiple Plots

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]


plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title('Plot One')



plt.subplot(2,2,2)
plt.plot(x,y2,color='blue')
plt.title('Plot Two')



plt.subplot(2,2,3)
plt.plot(x,y1,color='yellow')
plt.title('Plot One')



plt.subplot(2,2,4)
plt.plot(x,y2,color='red')
plt.title('Plot Two')
```

```python
### Bar plot

categories=['A','B','C','D','E']
values=[5,7,3,8,6]

###Create a bar plot

plt.bar(categories,values,color='purple')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Basic Bar Plot')
plt.show()
```

```python
###Histogram -- used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin



data=[1,2,2,2,3,3,4,5,4,5,7,8,8,8,8,9,2,2,2,1,1]

##creating a histogram
plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
##Creating a scatter plot

x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.scatter(x,y,color='blue',marker='*')
```

```python
##Pie chart

labels=['A','B','C','D']
sizes=[30,20,40,10]
colors=['gold','yellowgreen','lightcoral','lightskyblue']
explode=(0.2,0,0,0)

##create a pie chart
plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct='%1.1f%%',shadow=True)
```

```python
###Sales data visualization
import pandas as pd
sales_data=pd.read_csv('sales_data.csv')
sales_data.head(5)
```

```python
sales_data.info()
```

```python
### Plot total sales by product

total_sales_by_product=sales_data.groupby('sub_category')['profit'].sum()
```

```python
total_sales_by_product
```

```python
total_sales_by_product.plot(kind='bar',color='teal')
```

```python
## PLot sales trend over time

sales_trend= sales_data.groupby('ship_date')['profit'].sum().reset_index()
sales_trend
```

```python
plt.plot(sales_trend['ship_date'],sales_trend['profit'])
```
