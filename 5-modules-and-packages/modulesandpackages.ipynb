### Importing modules in python: modules and packages

#### modules and packages help organize and reuse code

```python
import math

math.sqrt(16)
```

```python
from math import sqrt,pi

print(sqrt(25))
print(pi)
```

```python
import numpy as np

np.array([1,2,3,4,5,7])
```

```python
from math import *
sqrt(16)
```

```python
from package.maths import addition

addition(2,3)
```

```python
import os
print(os.getcwd())
```
